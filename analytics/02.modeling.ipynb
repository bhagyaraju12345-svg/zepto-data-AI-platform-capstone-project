```python
import pandas as pd

df = pd.read_csv("titanic.csv")
```

```python
# Part B
# Task 7 Train Test Split

from sklearn.model_selection import train_test_split

# Classification target
X = df.drop(columns=["survived"])
y = df["survived"]

# Check class balance
print("Class balance:")
print(y.value_counts())

print("\nClass proportions:")
print(y.value_counts(normalize=True))

# Stratified split BEFORE preprocessing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining shape:", X_train.shape)
print("Testing shape:", X_test.shape)

print("\nTraining class proportions:")
print(y_train.value_counts(normalize=True))

print("\nTesting class proportions:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
```

```python
# Numeric

num_features=["age","fare","sibsp","parch","pclass"]
```

```python
# Categorical

cat_features=["sex","embarked"]
```

```python
# Preprocessor

preprocessor=ColumnTransformer([

("num",
Pipeline([
("imputer",SimpleImputer(strategy="median")),
("scaler",StandardScaler())
]),
num_features),

("cat",
Pipeline([
("imputer",SimpleImputer(strategy="most_frequent")),
("encoder",OneHotEncoder(handle_unknown="ignore"))
]),
cat_features)

])
```

```python
# Task 9 Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
```

```python
# Decision Tree Plot

dt_pipeline = Pipeline([
    ("prep", preprocessor),
    ("model", DecisionTreeClassifier(random_state=42))
])

dt_pipeline.fit(X_train, y_train)

tree_model = dt_pipeline.named_steps['model']

transformed_features = dt_pipeline.named_steps['prep'].get_feature_names_out()

plt.figure(figsize=(20, 10))
plot_tree(
    tree_model,
    filled=True,
    feature_names=transformed_features,
    class_names=["Dead", "Survived"],
    max_depth=3
)
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

# Create a pipeline for the Random Forest Classifier
rf_pipeline = Pipeline([
    ("prep", preprocessor),
    ("model", RandomForestClassifier(oob_score=True, bootstrap=True, random_state=42))
])

# Initialize GridSearchCV
grid = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5, # Using 5-fold cross-validation
    scoring='accuracy', # Or 'roc_auc' or 'f1'
    n_jobs=-1, # Use all available cores
    verbose=1
)

# Fit GridSearchCV to the resampled training data
grid.fit(X_train, y_train)

print(grid.best_params_)

print(grid.best_estimator_.named_steps["model"].oob_score_)
```

```python
logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000, random_state=42))
])

tree_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(
        max_depth=5,
        random_state=42
    ))
])

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ))
])
```

```python
logistic_pipeline.fit(X_train, y_train)

tree_pipeline.fit(X_train, y_train)

rf_pipeline.fit(X_train, y_train)
```

```python
# 10. Evaluate All Three Models
# Import metrics:
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve
)
```

```python
# Create an evaluation function:

import matplotlib.pyplot as plt
import seaborn as sns

def evaluate_model(model, X_test, y_test, model_name):

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    cm = confusion_matrix(y_test, y_pred)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob)

    print(f"\n{'='*50}")
    print(model_name)
    print(f"{'='*50}")

    print("Confusion Matrix:")
    print(cm)

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"AUC      : {auc:.4f}")

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "AUC": auc
    }
```

```python
logistic_results = evaluate_model(
    logistic_pipeline,
    X_test,
    y_test,
    "Logistic Regression"
)

tree_results = evaluate_model(
    tree_pipeline,
    X_test,
    y_test,
    "Decision Tree"
)

rf_results = evaluate_model(
    rf_pipeline,
    X_test,
    y_test,
    "Random Forest"
)
```

```python
# Comparison Table
classification_results = pd.DataFrame([
    logistic_results,
    tree_results,
    rf_results
])

print(classification_results)
```

```python
# Confusion Matrices

models = {
    "Logistic Regression": logistic_pipeline,
    "Decision Tree": tree_pipeline,
    "Random Forest": rf_pipeline
}

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, model) in zip(axes, models.items()):

    predictions = model.predict(X_test)

    cm = confusion_matrix(y_test, predictions)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
```

```python
# ROC Curves

plt.figure(figsize=(8, 6))

for name, model in models.items():

    probabilities = model.predict_proba(X_test)[:, 1]

    fpr, tpr, _ = roc_curve(y_test, probabilities)

    auc = roc_auc_score(y_test, probabilities)

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC = {auc:.3f})"
    )

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curves")

plt.legend()

plt.show()
```

```python
# Metrics
# Residual plot
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
```

```python
# Define features (X_reg) and target (y_reg) for regression
# Exclude 'survived' (if it was the classification target) and 'fare' (which is the new target)
X_reg = df.drop(columns=["fare", "survived", "alive", "class", "who", "adult_male", "alone"])
y_reg = df["fare"]

# Split data for regression
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# --- Define new preprocessor specifically for regression task ---
# 'fare' is now the target, so it should not be in num_features
# 'embark_town' is also a categorical feature in X_reg and should be handled
num_features_reg = ["age", "sibsp", "parch", "pclass"]
cat_features_reg = ["sex", "embarked", "embark_town"] # Added 'embark_town'

preprocessor_reg = ColumnTransformer([
    ("num",
     Pipeline([
         ("imputer", SimpleImputer(strategy="median")),
         ("scaler", StandardScaler())
     ]),
     num_features_reg),

    ("cat",
     Pipeline([
         ("imputer", SimpleImputer(strategy="most_frequent")),
         ("encoder", OneHotEncoder(handle_unknown="ignore"))
     ]),
     cat_features_reg)
])
# --- End of new preprocessor definition ---


# Create a regression pipeline
regression_pipeline = Pipeline([
    ("prep", preprocessor_reg), # Use the new regression preprocessor
    ("model", LinearRegression())
])

# Train the regression model
regression_pipeline.fit(X_train_reg, y_train_reg)

# Make predictions on the test set
pred = regression_pipeline.predict(X_test_reg)

# Calculate residuals
residuals = y_test_reg - pred

# Plot residuals
plt.figure(figsize=(10, 6))
plt.scatter(pred, residuals)
plt.axhline(0, color='red', linestyle='--')
plt.title('Residual Plot')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.grid(True)
plt.show()

# Also print some regression metrics
print(f"Mean Squared Error: {mean_squared_error(y_test_reg, pred):.2f}")
print(f"R-squared: {r2_score(y_test_reg, pred):.2f}")
```

```python
# Regression Metrics

# Regression Metrics

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

# Use the variables that were actually created in the regression section
y_reg_test = y_test_reg
y_reg_pred = pred

mae = mean_absolute_error(y_reg_test, y_reg_pred)
rmse = np.sqrt(mean_squared_error(y_reg_test, y_reg_pred))
r2 = r2_score(y_reg_test, y_reg_pred)

print(f"MAE: {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R^2: {r2:.4f}")
```

```python
# Adjusted R²

n = len(y_reg_test)
p = len(
    regression_pipeline.named_steps["prep"].get_feature_names_out()
)

if (n - p - 1) == 0:
    print("Adjusted R² is undefined because n - p - 1 = 0.")
else:
    adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - p - 1))
    print(f"Adjusted R²: {adjusted_r2:.4f}")
```

```python
print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
print("Adjusted R²:", adjusted_r2)
```
